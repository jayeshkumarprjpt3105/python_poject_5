```python

class Employee:
    def __init__(self, name, age, emp_id, salary):
        self.__emp_id = emp_id
        self.__salary = salary   # Private Variable
        self.name = name
        self.age = age 

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary >= 0:
            self.__salary = salary
        else:
            print("Salary cannot be negative!")

    def display(self):
        return f"Name: {self.name}, Age: {self.age}, ID: {self.__emp_id}, Salary: ₹{self.__salary}" 

class Manager(Employee):
    def __init__(self, name, age, emp_id, salary, department):
        super().__init__(name, age, emp_id, salary)
        self.department = department
    
    def display(self):
        return f"{super().display()}, Department: {self.department}"


class Developer(Employee):
    def __init__(self, name, age, emp_id, salary, language):
        super().__init__(name, age, emp_id, salary)
        self.language = language
    
    def display(self):
        return f"{super().display()}, Language: {self.language}"


# Lists to store data
all_employees = []
all_managers = []
all_developers = []

def show_menu():
   
    print("1. Create Employee")
    print("2. Create Manager")
    print("3. Create Developer")
    print("4. Show All")
    print("5. Exit")



def create_employee():
    print("\n--- Create Employee ---")
    name = input("Enter Name: ")
    age = int(input("Enter Age: "))
    emp_id = input("Enter Employee ID: ")
    salary = float(input("Enter Salary: "))
    e = Employee(name, age, emp_id, salary)
    all_employees.append(e)
    print(f"Employee Created: {e.display()}")

def create_manager():
    print("\n--- Create Manager ---")
    name = input("Enter Name: ")
    age = int(input("Enter Age: "))
    emp_id = input("Enter Employee ID: ")
    salary = float(input("Enter Salary: "))
    dept = input("Enter Department: ")
    m = Manager(name, age, emp_id, salary, dept)
    all_managers.append(m)
    print(f" Manager Created: {m.display()}")

def create_developer():
    print("\n--- Create Developer ---")
    name = input("Enter Name: ")
    age = int(input("Enter Age: "))
    emp_id = input("Enter Employee ID: ")
    salary = float(input("Enter Salary: "))
    lang = input("Enter Programming Language: ")
    d = Developer(name, age, emp_id, salary, lang)
    all_developers.append(d)
    print(f" Developer Created: {d.display()}")

def show_all():
    print("\n" + "="*50)
    print("  ALL RECORDS")
    print("="*50)
    
    
    
    print(f"\n Employees ({len(all_employees)}):")
    for e in all_employees:
        print(f"  • {e.display()}")
    
    print(f"\n Managers ({len(all_managers)}):")
    for m in all_managers:
        print(f"  • {m.display()}")
    
    print(f"\n Developers ({len(all_developers)}):")
    for d in all_developers:
        print(f"  • {d.display()}")
    
    total =len(all_employees) + len(all_managers) + len(all_developers)
    print(f"\n Total Objects: {total}")


while True:
    show_menu()
    choice = input("\nEnter your choice: ")
    
    
    if  choice == "1":
        create_employee()
    elif choice == "2":
        create_manager()
    elif choice == "3":
        create_developer()
    elif choice == "4":
        show_all()
    elif choice == "":
        print("\n Exiting... Goodbye!")
        break
    else:
        print(" Invalid choice!")
    
    print("\n" + "-"*50)
```
